```python
#Task 1
transactions = [
    {"id": 1, "product": "Laptop", "price": 1200, "quantity": 2, "region": "EU"},
    {"id": 2, "product": "Phone", "price": 800, "quantity": -1, "region": "US"},
    {"id": 3, "product": "Tablet", "price": 0, "quantity": 5, "region": "EU"},
    {"id": 4, "product": "Monitor", "price": 350, "quantity": 3, "region": "US"},
    {"id": 5, "product": "Laptop", "price": 1200, "quantity": 1, "region": "EU"},
    {"id": 6, "product": "Keyboard", "price": 75, "quantity": 10, "region": "APAC"},
]
```

```python
revenue_by_region = {}
for t in transactions:
    if t["price"] != 0 and t["quantity"] >= 0:
        t["total"] = t["price"] * t["quantity"]
        region = t["region"]
        if region in revenue_by_region:
            revenue_by_region[region] += t["total"]
        else:
            revenue_by_region[region] = t["total"]
            
```

```python
best_region =  ""
highest_revenue = 0
for r, rev in revenue_by_region.items():
    if rev > highest_revenue:
        highest_revenue = rev
        best_region = r
print(f"The region with highest revenue is {best_region} with {highest_revenue}.")
```

```python
#Task 2:
sales = [1200, 980, 1540, 870, 2010, 1650, 1300, 1100, 950, 1780, 2200, 1430, 1600, 1900]
def summarise_sales(sales):
    total_sales = sum(sales)
    avg_sales = round(total_sales/len(sales), 2)
    best_day = sales.index(max(sales)) + 1
    worst_day = sales.index(min(sales)) + 1
    count_above =  0
    for value in sales:
        if value > avg_sales:
            count_above += 1
    return {
        "total": total_sales,
        "average": avg_sales,
        "best_day": worst_day,
        "above_average_count": count_above
    }
    
```

```python
result = summarise_sales(sales)
print(result)
```

```python
#Task 4
class Portfolio:
    def __init__(self, owner):
        self.owner = owner
        self.assets = {}  # Dictionary to store assets: {ticker: [(quantity, price_per_unit), ...]}
    
    def add_asset(self, ticker, quantity, price_per_unit):
        """Add an asset to the portfolio or update existing asset quantity"""
        if ticker in self.assets:
            self.assets[ticker].append((quantity, price_per_unit))
        else:
            self.assets[ticker] = [(quantity, price_per_unit)]
    
    def total_value(self):
        """Calculate the total value of all assets in the portfolio"""
        total = 0
        for ticker, purchases in self.assets.items():
            for quantity, price in purchases:
                total += quantity * price
        return total
    
    def most_valueable(self):
        """Return the ticker of the most valuable asset"""
        max_value = 0
        max_ticker = None
        
        for ticker, purchases in self.assets.items():
            ticker_value = sum(quantity * price for quantity, price in purchases)
            if ticker_value > max_value:
                max_value = ticker_value
                max_ticker = ticker
                
        return max_ticker
    
    def __str__(self):
        """String representation of the portfolio"""
        result = f"Portfolio for {self.owner}:\n"
        for ticker, purchases in self.assets.items():
            total_quantity = sum(quantity for quantity, _ in purchases)
            total_value = sum(quantity * price for quantity, price in purchases)
            result += f"{ticker}: {total_quantity} shares, total value: ${total_value}\n"
        return result

# demonstrate
my_portfolio = Portfolio("CHAU")
# add 3 assets
my_portfolio.add_asset("VF", 10, 150)
my_portfolio.add_asset("GG", 2, 2800)
my_portfolio.add_asset("OR", 5, 100)

my_portfolio.add_asset("VF", 5, 160)

print(" Test Portfolio Method")
print(my_portfolio)
print(f"Total Portfolio Value: {my_portfolio.total_value()}")
print(f"Most Valuable Asset Ticker: {my_portfolio.most_valueable()}")
```

```python
#Task 5
from datetime import datetime
orders = [
    ("2025-01-15", "Laptop"), ("2025-01-22", "Phone"),
    ("2025-02-03", "Laptop"), ("2025-02-14", "Tablet"),
    ("2025-03-01", "Phone"), ("2025-03-10", "Phone"),
    ("2025-03-22", "Laptop"), ("2025-04-05", "Monitor"),
]
order_dates = []
for date_str, product in orders:
    d = datetime.fromisoformat(date_str)
    order_dates.append(d)
earliest = min(order_dates)
latest = max(order_dates)
diff = latest - earliest

print(f"Earliest order: {earliest}")
print(f"Latest order: {latest}")
print(f"Number of days between them: {diff.days} days")


```

a) Create a DataFrame with columns: date, visitors, conversion_rate, and avg_order_value. Set date as the index.
b) Compute two new columns: conversions (= visitors × conversion_rate, rounded to the nearest integer) and daily_revenue (= conversions × avg_order_value).


```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(seed=123)
days = pd.date_range(start="2025-01-06", periods=28, freq="D")
visitors = rng.integers(200, 600, size=28)
conversion_rate = rng.uniform(0.02, 0.08, size=28).round(4)
avg_order_value = rng.normal(50, 10, size=28).round(2)

df_traffic = pd.DataFrame({
    "date": days,
    "visitors": visitors,
    "conversion_rate": conversion_rate,
    "avg_order_value": avg_order_value
})

df_traffic = df_traffic.set_index("date")
print("a) Dataframe after set date as the index ( print the first 5 days ):")
print(df_traffic.head())
print("-" *50)

df_traffic["conversions"] = (df_traffic["visitors"] * df_traffic["conversion_rate"]).round().astype(int)
df_traffic["daily_revenue"] = df_traffic["visitors"] * df_traffic["avg_order_value"]

print("b) DataFrame after computing two new columns:")
print(df_traffic.head())
```
